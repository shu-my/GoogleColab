```python
#!/usr/bin/env python3
import jax
import jax.numpy as jnp
import numpy as np
jax.config.update("jax_enable_x64", True)

def _stable_switch(distance, r0, n_exp, m_exp):
    x = distance / r0
    numerator = 1.0 - x**n_exp
    denominator = 1.0 - x**m_exp
    near_r0 = jnp.abs(denominator) < 1.0e-7
    safe_denominator = jnp.where(near_r0, 1.0, denominator)
    value = numerator / safe_denominator
    dx = x - 1.0
    limit = (n_exp / m_exp) * (1.0 + 0.5 * (n_exp - m_exp) * dx)
    return jnp.where(near_r0, limit, value)

# 1. x=1 での値
r0, n_exp, m_exp = 0.8, 6, 12
print(_stable_switch(jnp.asarray(r0), r0, n_exp, m_exp))  # → 0.5 == n/m

# 2. 分岐境界(|denominator|=1e-7)をまたいでも値が連続か
# 3. x=1から離れた場所で直接式 (1-x^n)/(1-x^m) と一致するか
# 4. jax.grad が x=1 でも NaN にならず、sympyで求めた真の解析的微分値と一致するか
# 5. _coordination_number 全体としての和が幾何学的に妥当か、
#    最小image(周期境界)をまたいだ粒子でも同じ値になるか
```
